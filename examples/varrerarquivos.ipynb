```python
import os

def export_backend_code(output_file, root_dir=".", exclude_list=None, file_types=None):
    """
    Exporta o código do backend recursivamente para um arquivo, incluindo o caminho relativo e o conteúdo
    dos arquivos do tipo especificado, com a possibilidade de excluir arquivos ou pastas.

    Parâmetros:
        output_file (str): Caminho para o arquivo de saída.
        root_dir (str): Diretório raiz para iniciar a varredura.
        exclude_list (list): Lista de arquivos ou pastas a serem ignorados.
        file_types (list): Lista de extensões de arquivo a serem consideradas (e.g., ['.py']).

    Retorna:
        None. Apenas grava os resultados no arquivo de saída.
    """
    if exclude_list is None:
        exclude_list = []

    if file_types is None:
        file_types = [".py"]  # Por padrão, considera apenas arquivos Python

    # Converter root_dir para absoluto
    root_dir = os.path.abspath(root_dir)

    # Converter exclude_list para caminhos absolutos
    exclude_list = [os.path.abspath(os.path.join(root_dir, path)) for path in exclude_list]

    with open(output_file, "w", encoding="utf-8") as out_file:
        for root, dirs, files in os.walk(root_dir):
            # Remover pastas excluídas da varredura
            dirs[:] = [d for d in dirs if os.path.join(root, d) not in exclude_list]
            temp = [os.path.join(root, d) for d in dirs ]
            print(temp)

            for file in files:
                file_path = os.path.join(root, file)
                print("Incluindo arquivo:", file_path)

                # Ignorar arquivos excluídos ou não correspondentes ao tipo
                if file_path in exclude_list or not any(file.endswith(ext) for ext in file_types):
                    print("Excluindo arquivo:", file_path)
                    continue

                # Exibir caminho relativo e conteúdo do arquivo
                relative_path = os.path.relpath(file_path, root_dir)
                header = f"""\n
                ****************************************
                ********** {relative_path} **********
                ****************************************\n"""
                out_file.write(header)
                with open(file_path, "r", encoding="utf-8") as f:
                    out_file.write(f.read())

if __name__ == "__main__":
    # Configurações Backend
    OUTPUT_FILE = r"D:\Nuvem\ghdaru\OneDrive\030_DOUTORADO\120_TESE\130_TESEGIT\activetextclassification\examples\active.txt"  # Nome do arquivo de saída
    ROOT_DIR = r"D:\Nuvem\ghdaru\OneDrive\030_DOUTORADO\120_TESE\130_TESEGIT\activetextclassification\activetextclassification" # Diretório raiz da aplicação
    EXCLUDE_LIST = ["./activetextclassification/__pycache__", 
                    "./app/rag/__pycache__",
                    "./app/storages/__pycache__",
                    "./app/chats/__pycache__",
                    "./app/api/__pycache__", 
                    "./app/connections/__pycache__",
                    "./app/core/__pycache__", 
                    "./app/models/__pycache__", 
                    "./app/services/__pycache__",
                    "./app/files/__pycache__",
                    "./app/zold_api",
                    "./app/zold_connections",
                    "./app/zold_services",
                    "./app/zold_models",
                    "./app/rag",
                    "./app/storages",
                    "./app/core",
                    "./app/files/rou"                    
                    ]  # Exemplo de exclusões
    FILE_TYPES = [".py"]  # Tipos de arquivo a incluir
    # Configurações
    # OUTPUT_FILE = "frontend_export.txt"  # Nome do arquivo de saída
    # ROOT_DIR = r"D:\Nuvem\ghdaru\OneDrive\010_PROJETOS\3150_INFORMAAI\djangocourse\backendfast"  # Diretório raiz da aplicação
    # EXCLUDE_LIST = [ "firebase"]  # Exemplo de exclusões
    # FILE_TYPES = [".py"]  # Tipos de arquivo a incluir



    # Executar exportação
    export_backend_code(OUTPUT_FILE, ROOT_DIR, EXCLUDE_LIST, FILE_TYPES)
```
