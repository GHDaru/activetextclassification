```python
import json
import argparse
import os

def notebook_to_py(notebook_path, output_path=None):
    """
    Converte um arquivo Jupyter Notebook (.ipynb) para um script Python (.py),
    unificando todas as células de código. Linhas de "mágica" do IPython e
    comandos de shell serão comentados.

    Args:
        notebook_path (str): O caminho para o arquivo .ipynb de entrada.
        output_path (str, optional): O caminho para o arquivo .py de saída.
                                     Se None, o nome do arquivo .py será o mesmo
                                     do notebook, mas com a extensão .py, salvo
                                     no mesmo diretório.
    """
    if not os.path.exists(notebook_path):
        print(f"Erro: Arquivo notebook '{notebook_path}' não encontrado.")
        return

    if not notebook_path.lower().endswith(".ipynb"):
        print(f"Erro: O arquivo de entrada '{notebook_path}' não parece ser um Jupyter Notebook (.ipynb).")
        return

    if output_path is None:
        base_name = os.path.splitext(notebook_path)[0]
        output_path = base_name + ".py"
    elif not output_path.lower().endswith(".py"):
        print(f"Aviso: O caminho de saída '{output_path}' não termina com .py. A extensão .py será usada.")
        base_name_out = os.path.splitext(output_path)[0]
        output_path = base_name_out + ".py"


    try:
        with open(notebook_path, 'r', encoding='utf-8') as f:
            notebook_content = json.load(f)
    except json.JSONDecodeError:
        print(f"Erro: Não foi possível decodificar o JSON do arquivo '{notebook_path}'. O arquivo pode estar corrompido.")
        return
    except Exception as e:
        print(f"Erro ao abrir ou ler o arquivo '{notebook_path}': {e}")
        return

    all_code_lines = []
    cell_separator = "\n# ------------------ Célula Original ------------------\n"

    for i, cell in enumerate(notebook_content.get('cells', [])):
        if cell.get('cell_type') == 'code':
            source_lines = cell.get('source', [])
            if not source_lines:  # Pula células de código vazias
                continue

            all_code_lines.append(f"\n# Código da Célula {i+1}\n")
            
            processed_cell_lines = []
            for line_content in source_lines:
                stripped_line = line_content.lstrip() # Remove espaços à esquerda para checar o início
                # Comenta magias do IPython e comandos de shell
                if stripped_line.startswith(('%', '!')) or \
                   'get_ipython().run_line_magic' in stripped_line or \
                   'get_ipython().system' in stripped_line:
                    processed_cell_lines.append("# " + line_content)
                else:
                    processed_cell_lines.append(line_content)
            
            # Junta as linhas da célula, mantendo os newlines originais dentro da célula
            all_code_lines.append("".join(processed_cell_lines))
            # Adiciona um separador visual, mas garante que não haja múltiplos newlines vazios
            if not ("".join(processed_cell_lines)).strip().endswith("\n\n"):
                 all_code_lines.append("\n")


    final_script_content = "".join(all_code_lines).strip() # Remove newlines extras no início/fim

    try:
        with open(output_path, 'w', encoding='utf-8') as f:
            f.write(final_script_content)
        print(f"Notebook '{notebook_path}' convertido com sucesso para '{output_path}'")
    except Exception as e:
        print(f"Erro ao salvar o arquivo Python '{output_path}': {e}")

# if __name__ == "__main__":
#     parser = argparse.ArgumentParser(
#         description="Converte um Jupyter Notebook (.ipynb) para um script Python (.py)."
#     )
#     parser.add_argument(
#         "notebook_file",
#         help="Caminho para o arquivo .ipynb de entrada."
#     )
#     parser.add_argument(
#         "-o", "--output",
#         help="Caminho para o arquivo .py de saída (opcional). Se não fornecido, "
#              "o arquivo .py terá o mesmo nome do notebook e será salvo no mesmo diretório."
#     )

#     args = parser.parse_args()

    # notebook_to_py(args.notebook_file, args.output)
source = r'D:\Nuvem\ghdaru\OneDrive\030_DOUTORADO\120_TESE\130_TESEGIT\activetextclassification\examples\oraculo.ipynb'
dest = '.\oraculo'
notebook_to_py(source, dest)  # Exemplo de uso direto
```
