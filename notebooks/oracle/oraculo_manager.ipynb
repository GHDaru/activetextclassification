```python
# ==============================================================================
#           ORÁCULO - GERENCIADOR E ANALISADOR DE EXPERIMENTOS
# ==============================================================================
# Este notebook executa três etapas principais:
# 1. ETAPA DE EXECUÇÃO: Roda os experimentos definidos no arquivo JSON,
#    gerando os dados brutos em arquivos CSV (`control_results.csv` e `api_calls.csv`).
# 2. ETAPA DE ANÁLISE: Lê os arquivos CSV gerados para criar relatórios
#    consolidados em formato Excel (sumário e erros).
# 3. ETAPA DE PÓS-PROCESSAMENTO: Converte o arquivo de resultados principal
#    de CSV para Excel para facilitar a análise manual.

import os
import sys
import pandas as pd

# --- 1. CONFIGURAÇÃO DO AMBIENTE ---
# Adiciona o diretório raiz do projeto ao path para permitir a importação.
try:
    # Importa a classe principal e as funções de relatório
    from activetextclassification.oraculo.experiment_manager import (
        OraculoExperimentManager, 
        generate_summary_report, 
        generate_error_report,
        convert_csv_to_excel
    )
    print("Módulos importados com sucesso.")
except ImportError:
    print("Módulos não encontrados. Tentando ajustar o sys.path...")
    # Assume que o notebook está em /tese/activetextclassification/examples
    # Sobe um nível para /tese/activetextclassification/
    project_root = os.path.abspath(os.path.join(os.getcwd(), '../..'))
    if project_root not in sys.path:
        print(f"Adicionando diretório raiz ao path: {project_root}")
        sys.path.insert(0, project_root)
    
    from activetextclassification.oraculo.experiment_manager import (
        OraculoExperimentManager, 
        generate_summary_report, 
        generate_error_report,
        convert_csv_to_excel
    )
    print("Módulos importados com sucesso após ajuste do path.")


# --- 2. DEFINIÇÃO DOS CAMINHOS DOS ARQUIVOS ---
print("\n--- CONFIGURANDO OS CAMINHOS DOS ARQUIVOS ---")

base_path = r"C:\Users\ghdar\OneDrive\1000_tese\activetextclassification"

# Caminhos para os arquivos de dados brutos (CSV)
control_csv_path = os.path.join(base_path, "data_oraculo", "control_results.csv")
calls_csv_path = os.path.join(base_path, "data_oraculo", "api_calls.csv")

# Caminhos para os arquivos de relatório de saída (sem extensão)
summary_output_path = os.path.join(base_path, "data_oraculo", "report_summary")
error_output_path = os.path.join(base_path, "data_oraculo", "report_errors")
# Novo caminho para o Excel convertido (sem extensão)
control_excel_output_path = os.path.join(base_path, "data_oraculo", "control_results_export") 

# Caminhos de configuração
experiments_json_path = os.path.join(base_path, "data_oraculo", "experimentos_oraculo.json")
data_csv_path = os.path.join(base_path, "data_splits_cache", "dri_coldstart_selection_details_log.csv")
labels_json_path = os.path.join(base_path, "data_splits_cache", "all_possible_labels_original.json")


# ==============================================================================
#                       ETAPA 1: EXECUÇÃO DOS EXPERIMENTOS
# ==============================================================================

# Defina se você quer rodar a etapa de execução.
# Mude para False se você só quiser gerar relatórios de dados existentes.
EXECUTAR_EXPERIMENTOS = True

if EXECUTAR_EXPERIMENTOS:
    print("\n--- INICIANDO A ETAPA 1: EXECUÇÃO DOS EXPERIMENTOS ---")

    manager = OraculoExperimentManager(
        control_path=control_csv_path,
        calls_path=calls_csv_path,
        experiments_path=experiments_json_path,
        data_path=data_csv_path,
        labels_path=labels_json_path
    )

    print(manager.lista_categorias_str_mestra)

    manager.run_experiments()

    print("\n--- ETAPA DE EXECUÇÃO CONCLUÍDA ---")
else:
    print("\n--- ETAPA 1: EXECUÇÃO DOS EXPERIMENTOS (PULADA) ---")


# ==============================================================================
#                  ETAPA 2: GERAÇÃO DOS RELATÓRIOS ANALÍTICOS
# ==============================================================================

# Defina se você quer rodar a etapa de análise.
EXECUTAR_ANALISE = True

if EXECUTAR_ANALISE:
    print("\n--- INICIANDO A ETAPA 2: GERAÇÃO DOS RELATÓRIOS (Excel) ---")

    if os.path.exists(control_csv_path):
        # Gera o relatório de sumário
        generate_summary_report(
            control_csv_path=control_csv_path,
            calls_csv_path=calls_csv_path,
            output_path_no_ext=summary_output_path,
            output_format='excel'
        )

        # Gera o relatório de erros
        generate_error_report(
            control_csv_path=control_csv_path,
            output_path_no_ext=error_output_path,
            output_format='excel'
        )
    else:
        print(f"AVISO: Arquivo de controle '{control_csv_path}' não encontrado. Relatórios analíticos não foram gerados.")
else:
    print("\n--- ETAPA 2: GERAÇÃO DOS RELATÓRIOS (PULADA) ---")


# ==============================================================================
#                ETAPA 3: PÓS-PROCESSAMENTO E CONVERSÃO
# ==============================================================================

# Defina se você quer rodar a etapa de conversão.
EXECUTAR_CONVERSAO_EXCEL = True

if EXECUTAR_CONVERSAO_EXCEL:
    print("\n--- INICIANDO A ETAPA 3: CONVERSÃO DE CSV PARA EXCEL ---")

    if os.path.exists(control_csv_path):
        # Chama a função para converter o arquivo de resultados principal para Excel
        convert_csv_to_excel(
            csv_path=control_csv_path,
            excel_path_no_ext=control_excel_output_path
        )
    else:
        print(f"AVISO: Arquivo de controle '{control_csv_path}' não encontrado. Conversão para Excel não foi realizada.")
else:
    print("\n--- ETAPA 3: CONVERSÃO DE CSV PARA EXCEL (PULADA) ---")


print("\n--- PROCESSO COMPLETO FINALIZADO ---")
```
